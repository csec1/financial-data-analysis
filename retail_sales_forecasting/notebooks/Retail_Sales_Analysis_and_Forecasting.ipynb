```r
# Load required libraries
library(readxl)    # For reading the Excel file
library(dplyr)     # For data manipulation
library(ggplot2)   # For plotting
```

```r
# Set the file path for the Excel file
file_path <- "/home/cybersec/python_projects/financial-data-analysis/retail_sales_forecasting/notebooks/Online_Retail.xlsx"

# Load the data from the Excel file
data <- read_excel(file_path)

# View the first few rows of the data to confirm it has loaded
head(data)
```

```r
# Data cleaning: Remove rows with missing InvoiceNo or CustomerID
data_clean <- data %>%
  filter(!is.na(InvoiceNo), !is.na(CustomerID))

# Convert InvoiceDate to Date type
data_clean$InvoiceDate <- as.Date(data_clean$InvoiceDate, format="%d/%m/%Y")

# Calculate total sales (Quantity * UnitPrice)
data_clean$TotalSales <- data_clean$Quantity * data_clean$UnitPrice

# View the first few rows of the cleaned data
head(data_clean)



```

```r
# Summarize data by country (top 10 countries by total sales)
country_sales <- data_clean %>%
  group_by(Country) %>%
  summarise(TotalSales = sum(TotalSales)) %>%
  arrange(desc(TotalSales)) %>%
  top_n(10)

# View the summary of sales by country
head(country_sales)
```

```r
# Plot 1: Bar plot for total sales by country (Top 10)
ggplot(country_sales, aes(x = reorder(Country, TotalSales), y = TotalSales, fill = Country)) +
  geom_bar(stat = "identity") +
  coord_flip() +
  theme_minimal() +
  labs(title = "Top 10 Countries by Total Sales", x = "Country", y = "Total Sales")
```

```r
# Summarize total sales by month
monthly_sales <- data_clean %>%
  mutate(Month = format(InvoiceDate, "%Y-%m")) %>%
  group_by(Month) %>%
  summarise(TotalSales = sum(TotalSales))

# View the monthly sales summary
head(monthly_sales)
```

```r
# Plot 2: Time series plot for total sales over time (monthly)
ggplot(monthly_sales, aes(x = Month, y = TotalSales)) +
  geom_line(color = "blue") +
  theme_minimal() +
  labs(title = "Total Sales Over Time (Monthly)", x = "Month", y = "Total Sales") +
  theme(axis.text.x = element_text(angle = 90, hjust = 1))
```

```r
# Plot 3: Histogram of sales distribution
ggplot(data_clean, aes(x = TotalSales)) +
  geom_histogram(binwidth = 5, fill = "lightblue", color = "black") +
  theme_minimal() +
  labs(title = "Distribution of Total Sales", x = "Total Sales", y = "Frequency")
```
